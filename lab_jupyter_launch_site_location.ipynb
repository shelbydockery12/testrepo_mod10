```python

# Apply a function to check the value of `class` column
# If class=1, marker_color value will be green
# If class=0, marker_color value will be red
color_list = []
def marker_color(value):
    marker_color_list = []
    for i in value:
        if i == 1:
            marker_color = 'green'
            marker_color_list.append(marker_color)
        elif i == 0:
            marker_color = 'red'
            marker_color_list.append(marker_color)
        else:
            print('no color available')
    color_list.append(marker_color_list)
    return color_list
marker_color(spacex_df['class'])

spacex_df['marker_color']=color_list[0]
print(spacex_df.head())
        
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```
